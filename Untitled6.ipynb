#Exercício 4

```python
#Solicite dois números e apresente os seguintes resultados: Média maior ou igual a 7: Aprovado;  Média entre 5 e 6,9: Recuperação; Média menor que 5: Reprovado.

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print(f"Média final: {media:.1f}")

if media >= 7.0:
 print(f"situação: Aprovado")
elif media >= 5.0 and media <= 6.9:
 print(f"situação: Recuperação")
else:
 print(f"situação: Reprovado")
#
```

#Exercício 10

```python
#Solicite dois números e uma operação matemática (+, -, * ou /). Realize a operação escolhida e exiba o resultado correspondente

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
operacao = input("Digite a operação desejada (+, -, * ou /): ")

if operacao == "+":
  resultado = num1 + num2
  print(f"{num1} + {num2} = {resultado}")
elif operacao == "-":
  resultado = num1 - num2
  print(f"{num1} - {num2} = {resultado}")
elif operacao == "*":
  resultado = num1 * num2
  print(f"{num1} * {num2} = {resultado}")
elif operacao == "/":
  if num2 != 0:
    resultado = num1 / num2
    print(f"{num1} / {num2} = {resultado}")
  else:
    print("Divisão por zero não é permitida")
else:
  print("Operação inválida")
#
```
