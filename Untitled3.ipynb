```python
# Escrever um jogo de adivinhação. O usuario terá 03 chances para adivinhar um numero de 0 a 10
# o programa dar uma mensagem se o palpite foi maior ou menor
# o programa avisar se está quente ou frio.

import random
numero_secreto =random.randint(1,10)
tentativa = 3
print("Jogo de adivinhação")
print("Tente adivinhar o numero entre 0 e 10. Você tem 3 chances.")

for i in range(tentativa):
  palpite =int(input(f"Tentiva {i+1}.Digite seu palpite: "))
  if palpite == numero_secreto:
    print(f"Parabéns! Você acertou! O número era {numero_secreto}.")
    break
    else:
      if palpite > numero_secreto:
        print("Errou! Seu palpite é Maior que o numero secreto.")
        else:
          print("Errou! Seu palpite é Menor que o numero secreto.")
          diferenca = abs(numero_secreto - palpite)
           if diferenca <= 2:
            print("Quente! está bem próximo do número secreto.")
             elif diferenca <= 4:
              print("Morno! está chegando quase lá...")
              else:
                print("frio! está bem longe!")

                print(f"Suas tentativas acabaram! O numero secreto era {numero_secreto}.")
```
