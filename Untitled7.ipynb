#Exercício 4

```python
#Solicite dois números e apresente os seguintes resultados: Média maior ou igual a 7: Aprovado;  Média entre 5 e 6,9: Recuperação; Média menor que 5: Reprovado.

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print(f"Média final: {media:.1f}")

if media >= 7.0:
 print(f"situação: Aprovado")
elif media >= 5.0 and media <= 6.9:
 print(f"situação: Recuperação")
else:
 print(f"situação: Reprovado")
#
```

#Exercício 5

```python
#Receba o valor de uma compra.Se o valor for maior ou igual a 100, aplique 10% de desconto e mostre o valor final. Caso contrário mostre o valor normal da compra

valor_compra = float(input("Digite o valor da compra: "))

if valor_compra >= 100:
  valor_final = valor_compra * 0.9
  print(f"Valor final com desconto: R${valor_final:.2f}")
else:
  print(f"Valor final sem desconto: R${valor_compra:.2f}")
#
```
