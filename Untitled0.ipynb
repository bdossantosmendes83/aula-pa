```python
# Escrever um programa de uma calculadora simples. O usuário irá digitar 02 números e escolherá uma operação.

# Entrada
print ("|-----------------CALCULADORA-------------|")
print ("|                                         |")
print ("|-----------------MENU--------------------|")
print ("|                                         |")
print ("|Opção 1 - Soma                           |")
print ("|Opção 2 - Subtração                      |")
print ("|Opção 3 - Multiplicação                  |")
print ("|Opção 4 - Divisão                        |")
print ("|----------------------------------------- |")

try:
    num1 = float(input("Digite o primeiro número: "))
    num2 = float(input("Digite o segundo número: "))
    opcao = input("Escolha a operação (1, 2, 3 ou 4): ")
 Processamento / Saída

    #==opcao '1':
       if resultado = num1 + num2
        print(f"{num1} + {num2} = {resultado}")
    elif opcao == '2':
        resultado = num1 - num2
        print(f"{num1} - {num2} = {resultado}")
    elif opcao == '3':
        resultado = num1 * num2
        print(f"{num1} * {num2} = {resultado}")
    elif opcao == '4':
        if num2 != 0:
            resultado = num1 / num2
            print(f"{num1} / {num2} = {resultado}")
        else:
            print("Erro: Divisão por zero não é permitida.")
    else:
        print("Opção inválida. Por favor, escolha uma opção entre 1 e 4.")
except ValueError:
    print("Entrada inválida. Por favor, digite apenas números para as operações.")
```

```python
# escrever um programa que defina cada dia da semana com um numero de 1 a 7

#entrada

numero = int(input("digite um numero de 1 a 7:"))

#processamento/saida
if numero == 1:
  print("segunda-feira")
elif numero == 2:
  print("terça-feira")
elif numero == 3:
  print("quarta-feira")
elif numero == 4:
  print("quinta-feira")
elif numero == 5:
  print("sexta-feira")
elif numero == 6:
  print("sabado")
elif numero == 7:
  print ("domingo")
else:
  print("Número inválido. Por favor, digite um número de 1 a 7.")
```
