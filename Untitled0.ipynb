Fazer um algoritmo e depois programar que recebe 4 notas do aluno e calcula a média aritmética e imprimi a média.

# entrada
ler nota1
ler nota2
ler nota3
ler nota4
#processamento
 calcular a média das notas

# saida
 escrever o total da média das 4 notas

```python
#variaveis que recebe os valores das notas

#entrada
nota_1 = 8
nota_2 = 9
nota_3 = 7
nota_4 = 3
```

```python
#processamento

media = (nota_1+nota_2+nota_3+nota_4)/4
```

```python
#saida

print (F"Total da media das 4 notas é {media}")
```
